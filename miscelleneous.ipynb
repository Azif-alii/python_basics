```python
class Employee:
  def __init__(self):
    self.name="Rahul"  #Public
    self._salary=5000  #Protected
    self._password="12345"  #Private

  def show_details(self):
    print(self.name)
    print(self._salary)
    print(self._password)

emp=Employee()

print(emp.name)    #works
print(emp._salary)  #Work,but discouraged
    #print(emp._password) #Error

emp.show_details()
```
